**1. Generate & Eksplorasi Dataset**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
from sklearn.metrics.pairwise import cosine_similarity
```

```python
np.random.seed(42)
produk = ['Roti', 'Selai', 'Susu', 'Sereal', 'Telur', 'Keju', 'Kopi', 'Gula', 'Teh', 'Mentega']

# Buat 50 transaksi, tiap transaksi berisi 2-5 produk
transaksi = []
for i in range(50):
    n_item = np.random.randint(2, 6)
    transaksi.append(list(np.random.choice(produk, n_item, replace=False)))

# Suntikkan pola: Roti sering bersama Selai
for i in range(0, 20):
    if 'Roti' in transaksi[i] and 'Selai' not in transaksi[i]:
        transaksi[i].append('Selai')

print('Contoh transaksi:', transaksi[:3])
print('Jumlah transaksi:', len(transaksi))
```

**2. One-Hot Encoding Transaksi**

```python
te = TransactionEncoder()
te_ary = te.fit(transaksi).transform(transaksi)
df = pd.DataFrame(te_ary, columns=te.columns_)

print(df.head())
```

**3. Frequent Itemset dengan Apriori**

```python
# Amati perubahan jumlah itemset berdasarkan min_support
for ms in [0.05, 0.1, 0.2]:
    freq = apriori(df, min_support=ms, use_colnames=True)
    print(f'min_support={ms}: {len(freq)} itemset ditemukan')

# Gunakan min_support yang menghasilkan jumlah itemset wajar
freq_items = apriori(df, min_support=0.1, use_colnames=True)
freq_items = freq_items.sort_values('support', ascending=False)

print("\nTop 10 Frequent Itemset:")
print(freq_items.head(10))
```

**4. Bentuk & Saring Aturan Asosiasi**

```python
rules = association_rules(freq_items, metric='confidence', min_threshold=0.5)
rules = rules[rules['lift'] > 1].sort_values('lift', ascending=False)

print("Aturan Asosiasi Terkuat (Top 10 berdasarkan Lift):")
print(rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']].head(10))
```

Analisis Langkah 4:
* Aturan terkuat (Lift tertinggi): Biasanya dipegang oleh pasangan Roti -> Selai (atau sebaliknya) karena ada manipulasi pola yang disuntikkan pada Langkah 1.
* Kewajaran Bisnis: Pola ini sangat masuk akal dalam dunia retail (Market Basket Analysis), karena roti dan selai merupakan barang komplementer (saling melengkapi).

**5. Rekomender Sederhana dengan Content-Based Filtering**

```python
katalog = pd.DataFrame({
    'produk': produk,
    'kategori': ['Bakery','Bakery','Dairy','Bakery','Dairy', 'Dairy','Minuman','Bumbu','Minuman','Dairy']
})

fitur = pd.get_dummies(katalog['kategori'])
sim_matrix = cosine_similarity(fitur)

def rekomendasi_serupa(nama_produk, top_n=3):
    idx = katalog.index[katalog['produk'] == nama_produk][0]
    skor = list(enumerate(sim_matrix[idx]))
    skor = sorted(skor, key=lambda x: x[1], reverse=True)
    skor = [s for s in skor if s[0] != idx][:top_n]
    return katalog.iloc[[i for i, _ in skor]]['produk'].tolist()

print('Mirip dengan Roti:', rekomendasi_serupa('Roti'))
```

**6. Perbandingan Kedua Pendekatan**

```python
produk_target = 'Roti'

# Dari association rules: cari consequents dari aturan yang antecedent-nya mengandung produk_target
rules_terkait = rules[rules['antecedents'].apply(lambda x: produk_target in x)]
print('Rekomendasi dari Association Rules:')
print(rules_terkait[['consequents', 'lift']].head())

print('\nRekomendasi dari Content-Based:', rekomendasi_serupa(produk_target))
```

### Diskusi & Analisis Hasil

#### 1. Konsistensi Hasil Rekomendasi
> **Kesimpulan:** Kedua pendekatan **tidak memberikan hasil yang sama (tidak konsisten)** karena cara kerja dan tujuannya berbeda.

* **Association Rules:** Fokus pada **perilaku belanja konsumen** di lapangan. Metode ini mencari hubungan produk pelengkap (*cross-selling*). Contohnya, orang yang membeli `Roti` cenderung membeli `Selai` karena kedua barang tersebut biasa dikonsumsi bersamaan.
* **Content-Based Filtering:** Murni melihat **kemiripan karakteristik atau atribut** pada katalog produk. Sistem merekomendasikan produk lain yang berada dalam satu kategori yang sama (misalnya sesama produk *Bakery* seperti `Sereal`), tanpa memedulikan apakah barang tersebut sering dibeli bersamaan atau tidak.

---

#### 2. Kondisi Penggunaan Terbaik
* **Content-Based Filtering cocok digunakan saat:**
  Menghadapi masalah *cold start*, yaitu kondisi ketika ada pengguna baru atau produk baru masuk ke toko yang belum memiliki riwayat transaksi atau interaksi sama sekali.
* **Association Rules / Collaborative Filtering cocok digunakan saat:**
  Data transaksi kolektif dari pelanggan sudah melimpah, sehingga pola hubungan antar-barang yang tidak kasat mata bisa terbaca dengan akurat secara statistik.
